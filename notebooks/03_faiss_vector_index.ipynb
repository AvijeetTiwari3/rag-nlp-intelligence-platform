**Objective:** Construct, serialize, and benchmark an inner-product dense index using Facebook AI Similarity Search (FAISS) over 4,039 passage vectors.


```python
import sys, os
sys.path.insert(0, '..')
import faiss
import numpy as np
import pandas as pd
from src.vector_store import FAISSVectorStore
from src.embedder import Embedder
from src.retriever import Retriever
```

```python
embedder = Embedder(model_name='sentence-transformers/all-MiniLM-L6-v2')
store = FAISSVectorStore(embedding_dim=embedder.get_embedding_dim())
store.load('../models/faiss_index.index', '../models/faiss_index_meta.parquet')
print(f'FAISS Index loaded successfully. Total Vectors: {store.count:,}')
```

```python
retriever = Retriever(embedder=embedder, vector_store=store)
query = 'What is the function of the human cardiovascular system?'
results = retriever.retrieve(query, top_k=3)

for i, r in enumerate(results, 1):
    print(f"[{i}] Score: {r['score']:.4f} | Title: {r['title']}")
    print(f"    Excerpt: {r['text'][:150]}...\n")
```
