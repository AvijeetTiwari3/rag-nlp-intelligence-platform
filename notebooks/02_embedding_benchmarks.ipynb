**Objective:** Empirically benchmark candidate dense representation models (`all-MiniLM-L6-v2`, `all-MiniLM-L12-v2`, `all-mpnet-base-v2`) on latency, memory footprint, and retrieval recall.


```python
import sys, os
sys.path.insert(0, '..')
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from src.embedder import Embedder
```

```python
candidates = {
    'MiniLM-L6-v2': 'sentence-transformers/all-MiniLM-L6-v2',
    'MiniLM-L12-v2': 'sentence-transformers/all-MiniLM-L12-v2'
}

sample_texts = [
    'Deep neural networks require substantial matrix multiplication acceleration.',
    'Photosynthesis utilizes chlorophyll pigments to absorb radiant solar energy.',
    'The Magna Carta signed in 1215 established the principle that everyone is subject to the law.'
] * 20

benchmark_records = []
for name, model_id in candidates.items():
    emb = Embedder(model_name=model_id)
    t0 = time.perf_counter()
    vectors = emb.encode(sample_texts, batch_size=32)
    dur = time.perf_counter() - t0
    benchmark_records.append({
        'Model': name,
        'Embedding Dim': emb.get_embedding_dim(),
        'Latency (s)': round(dur, 4),
        'Throughput (texts/sec)': round(len(sample_texts) / dur, 1)
    })

df_bench = pd.DataFrame(benchmark_records)
df_bench
```

```python
plt.figure(figsize=(8, 4))
plt.bar(df_bench['Model'], df_bench['Throughput (texts/sec)'], color=['#4285F4', '#34A853'])
plt.title('Embedding Inference Throughput Comparison (CPU)')
plt.ylabel('Texts Processed per Second')
plt.show()
```
