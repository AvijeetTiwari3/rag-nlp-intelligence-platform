**Objective:** Integrate context passages into high-grounding prompt templates with Google Gemini 1.5 Flash, measuring output faithfulness and latency.


```python
import sys, os
sys.path.insert(0, '..')
from src.generator import GeminiGenerator

api_key = os.environ.get('GEMINI_API_KEY', '')
if api_key:
    generator = GeminiGenerator(api_key=api_key)
    print('Gemini Generator initialized successfully.')
else:
    print('Please set GEMINI_API_KEY environment variable.')
```

```python
context_passages = [
    {'title': 'Alexander Fleming', 'text': 'In 1928, Alexander Fleming discovered penicillin from the mould Penicillium notatum.', 'score': 0.89}
]

result = generator.generate(
    question='Who discovered penicillin and in which year?',
    retrieved_passages=context_passages
)

print(f"Answer: {result['answer']}")
print(f"Latency: {result['latency_ms']} ms")
print(f"Sources: {result['sources']}")
```
