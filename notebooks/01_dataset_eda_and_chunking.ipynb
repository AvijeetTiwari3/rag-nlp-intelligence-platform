**Objective:** Ingest authentic Stanford Question Answering Dataset (SQuAD 2.0), analyze document lengths, sentence boundaries, answerable vs. unanswerable distribution, and establish semantic chunking boundaries.


```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
print('Packages loaded successfully.')
```

```python
corpus_path = '../data/processed/corpus_chunks.parquet'
golden_path = '../data/processed/eval_golden_set.parquet'

df_corpus = pd.read_parquet(corpus_path)
df_golden = pd.read_parquet(golden_path)

print(f'Corpus Chunks: {len(df_corpus):,}')
print(f'Golden Evaluation QA Pairs: {len(df_golden):,}')
```

### 1. Corpus Chunk Length Distribution
Understanding token/character distribution is critical to avoid truncating retrieval passages or exceeding embedding context windows (256 tokens for MiniLM).


```python
df_corpus['char_len'] = df_corpus['text'].str.len()
df_corpus['word_len'] = df_corpus['text'].str.split().str.len()

print(df_corpus[['char_len', 'word_len']].describe())
```

```python
plt.figure(figsize=(10, 4))
sns.histplot(df_corpus['word_len'], bins=40, kde=True, color='teal')
plt.title('Word Count Distribution Across 4,039 Wikipedia Passages')
plt.xlabel('Words per Passage Chunk')
plt.ylabel('Frequency')
plt.show()
```

### 2. Answerability Balance in Golden Set
SQuAD 2.0 uniquely tests model restraint by including unanswerable questions to detect over-confidence.


```python
print(df_golden['has_answer'].value_counts(normalize=True))
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df_golden, x='has_answer', palette='Set2')
plt.title('SQuAD 2.0 Golden Set: Answerable vs Unanswerable QA Pairs')
plt.xlabel('Has Valid Ground-Truth Answer')
plt.ylabel('Count')
plt.show()
```
